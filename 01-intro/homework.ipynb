## Matrix Operations

Select the columns ```vehicle_weight``` and ```model_year``` of the first seven rows of the cars from Asia. Get the underlying NumPy array (```X```). Compute the matrix multiplication between the transpose of ```X``` and ```X``` (```XTX```). Invert ```XTX```. Create an array ```y``` with values ```[1100, 1300, 800, 900, 1000, 1100, 1200]```. Multiply the inverse of ```XTX``` with the transpose of ```X``` and then multiply the result by ```y```. Sum all the elements of the result (Q7).

````python
X = df[df['origin'] == 'Asia'][['vehicle_weight','model_year']][:7].to_numpy()

X
````

````python
XTX = X.T.dot(X)

XTX
````

````python
XTX_inv = np.linalg.inv(XTX)

XTX_inv
````

````python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

y
````

````python
w = XTX_inv.dot(X.T).dot(y)

w
````

````python
sum(w)
````
